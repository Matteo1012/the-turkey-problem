```python
import random

def dynamic_turkey_simulation(num_batches=20, base_capacity=400, infringement_threshold=15):
    """
    Simulates 'The Turkey Problem' with a variable target capacity (n_i) 
    and state carryover (m): m + new_turkeys = n_i
    """
    print("🦃 THE TURKEY PROBLEM: DYNAMIC TARGET SIMULATOR")
    print(f"Base Target Capacity: {base_capacity}")
    print(f"Infringement Threshold (m_min): {infringement_threshold} old turkeys\n")

    carryover_m = 0  # Initial old turkeys carried over
    infringed_decisions = 0

    for batch in range(1, num_batches + 1):
        # Dynamic target capacity n_i fluctuates per batch due to operational variance
        target_n = int(random.gauss(base_capacity, 25))
        target_n = max(300, min(450, target_n))

        # Determine escaped/carried over turkeys
        if random.random() > 0.15:
            escaped_or_carried = random.randint(1, 4)
        else:
            escaped_or_carried = random.randint(12, 40)

        current_m = carryover_m + escaped_or_carried
        current_m = min(current_m, target_n)

        # Mass balance equation: m + new = n_i  =>  new = n_i - m
        new_turkeys_needed = target_n - current_m

        # Check if carryover infringes standard decision process
        if current_m < infringement_threshold:
            status = "✅ STABLE"
            action = f"Target n={target_n}. Ignore minor residue. Collect ~{new_turkeys_needed} new."
        else:
            status = "⚠️ INFRINGED"
            action = f"Target n={target_n}. Critical carryover m={current_m}! Must adjust to exactly {new_turkeys_needed} new."
            infringed_decisions += 1

        print(f"Batch {batch:02d} | Target (n): {target_n:03d} | Carryover (m): {current_m:02d} | {status:<12} | {action}")

        # Rollover a fraction of current_m to the next batch
        carryover_m = random.randint(0, min(5, current_m))

    print(f"\n--- Simulation Complete ---")
    print(f"Total Decisions Infringed: {infringed_decisions} out of {num_batches} batches.")

if __name__ == "__main__":
    dynamic_turkey_simulation()
```
