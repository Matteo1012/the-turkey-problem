```python
import random

def turkey_loading_simulation(num_batches=50, target_batch_size=400, anomaly_threshold=12):
    """
    Simulates the 'Turkey Problem' during agricultural loading operations.
    Focuses on the absolute error threshold (12-15 escapes) that skews spatial estimation.
    """
    print("THE TURKEY PROBLEM SIMULATOR")
    print(f"Target Batch Size: {target_batch_size} turkeys")
    print(f"Anomaly Threshold: {anomaly_threshold} escaped turkeys\n")

    compromised_decisions = 0

    for batch in range(1, num_batches + 1):
        
        if random.random() > 0.10:
            escaped = int(random.gauss(4.5, 1.5)) 
        else:
            escaped = random.randint(12, 16) 
            
        escaped = max(0, escaped)

        
        if escaped < anomaly_threshold:
            status = "STABLE"
            next_batch_estimation = target_batch_size
            action = f"Next batch estimated accurately (~{next_batch_estimation})"
        else:
            status = "OMPROMISED"
           
            next_batch_estimation = target_batch_size + random.randint(40, 60)
            action = f"Visual reference skewed! Next batch over-estimated (~{next_batch_estimation})"
            compromised_decisions += 1

        print(f"Batch {batch:02d} | Escaped: {escaped:02d} | {status:<15} | {action}")

    print(f"\n--- Simulation Complete ---")
    print(f"Total Compromised Decisions: {compromised_decisions} out of {num_batches} batches.")
    print(f"System Stability Rate: {((num_batches - compromised_decisions) / num_batches) * 100:.1f}%")

if __name__ == "__main__":
    turkey_loading_simulation()
```
